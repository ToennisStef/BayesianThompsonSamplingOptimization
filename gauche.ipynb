```python
from COSMOtherm.src.funcs import load_and_sort_solvents, list_files_with_extension
from COSMOtherm.src.BayesianOptimization import get_training_data
from COSMOtherm.Configfiles import Config

init_files = list_files_with_extension(Config.initialSamples_dir, 'tab')
solvents = load_and_sort_solvents(Config.solvents_fullpath)

train_X_init, train_Y_init, = get_training_data(init_files, solvents=solvents)
```

```python
import pandas as pd
import numpy as np

out_df = pd.read_csv("./output_data.csv")

# solvents.merge(out_df, how='left', left_on='COSMO_name', right_on='Compound 2')
out_df = out_df.rename(columns={"Compound 2": "COSMO_name"})
#add smiles from solvents to out_df
out_df["SMILES"] = out_df["COSMO_name"].map(solvents.set_index("COSMO_name")["SMILES"])


out_df['KV'] = out_df['Phase 2 x(3) output'] / out_df['Phase 1 x(3) output']
 
out_df.to_csv("./output_data_full.csv", index=False)
```
