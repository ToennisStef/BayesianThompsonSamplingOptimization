```python
import pandas as pd
from collections import defaultdict

Alcohols = pd.read_csv("COSMOtherm/Datafiles/Alcohols.txt")

# Group alcohols by their base name (e.g., "hexane") and create a structured dictionary

# Initialize a dictionary to group alcohols
grouped_alcohols = defaultdict(list)

for _, row in Alcohols.iterrows():
    parts = row['COSMO_name'].split('-')
    # If the name is like "1-hexanol", then parts[0] is "1" and parts[1] is "hexanol"
    if len(parts) > 1:
        major_group = parts[1]  # e.g. "hexanol"
    else:
        major_group = parts[0]
    grouped_alcohols[major_group].append(row.to_dict())

grouped_alcohols = dict(grouped_alcohols)

print(grouped_alcohols.keys())  # Check the groups
```
