```python
from COSMOtherm.src.funcs import load_and_sort_solvents
from COSMOtherm.src.Design_Matrix import build_design_matrix
from Archive.funcs2 import sort_solvents_df

import pandas as pd 

S1 = pd.read_csv(r"./COSMOtherm/Datafiles/Solvents.txt", delim_whitespace=True)
```

```python
sorted_df = sort_solvents_df(S1)  
```

```python
tC_range = [20, 40]  # [°C]
massc1_lacticacid_range = [5, 20]  # [g/L] mass concentration of lactic acid in the solution [250]
M_h2o = 18.01528  # [g/mol] molar mass of water
M_lacticacid = 90.078  # [g/mol] molar mass of lactic acid
rho_lacticacid = 1.209  # [g/mL] density of lactic acid
rho_h2o = 1.0  # [g/mL] density of water
c_p_h2o = 55.5  # [mol/L] concentration of pure water at 25 °C
c_p_lacticacid = rho_lacticacid / M_lacticacid * (10**3)  # [mol/L] concentration of pure lactic acid at 25 °C

# --- Helper Functions ---
def calculate_molar_fractions():

    c1_lacticacid_range = [
        massc1_lacticacid_range[0]/M_lacticacid,
        massc1_lacticacid_range[1]/M_lacticacid
    ]
    c1_h2o_range = [
        c_p_h2o - (c_p_h2o/c_p_lacticacid)*c1_lacticacid_range[0],
        c_p_h2o - (c_p_h2o/c_p_lacticacid)*c1_lacticacid_range[1]
    ]
    x1_lacticacid_range = [
        c1_lacticacid_range[0]/(c1_lacticacid_range[0]+c1_h2o_range[0]),
        c1_lacticacid_range[1]/(c1_lacticacid_range[1]+c1_h2o_range[1])
    ]
    return x1_lacticacid_range
    
```

```python
x1_lacticacid_range = calculate_molar_fractions()
solvents_ids = S1.index.tolist()
solvents_ids_range = [solvents_ids[0], solvents_ids[-1]]
    
```

```python
full_factorial = build_design_matrix(
    tC_range=tC_range, 
    x1_lacticacid_range=x1_lacticacid_range, 
    solvents=sorted_df,        
    reduction= 0,
)
```

```python
print(f"Full Factorial Design Matrix: {len(full_factorial)}")
```
