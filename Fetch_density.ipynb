```python
# Check the datatype of a specific column, for example 'density g/cm^3 @20°C'
print(solvents_df['density g/cm^3 @20°C'].dtype)

# Replace comma with dot and convert to float64
# Use regex to extract only the numeric part, replace comma with dot, and convert to float
solvents_df['density g/cm^3 @20°C'] = (
    solvents_df['density g/cm^3 @20°C']
    .str.extract(r'([\d,\.]+)', expand=False)  # extract numeric part
    .str.replace(',', '.', regex=False)        # replace comma with dot
    .astype('float64')
)
print(solvents_df['density g/cm^3 @20°C'].dtype)
```

```python
import requests

# Step 1: Resolve CAS number to CID
cas_number = "499-75-2"
url_cid = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{cas_number}/cids/JSON"

response_cid = requests.get(url_cid)
if response_cid.status_code == 200:
    cid_data = response_cid.json()
    cids = cid_data.get("IdentifierList", {}).get("CID", [])
    if cids:
        cid = cids[0]
        print(f"CID for CAS {cas_number}: {cid}")

        # Step 2: Use CID to get Density and other properties
        url_density = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{cid}/property/Density/JSON"
        response_density = requests.get(url_density)

        if response_density.status_code == 200:
            properties = response_density.json().get("PropertyTable", {}).get("Properties", [{}])[0]
            density = properties.get("Density", None)
            if density:
                print(f"Density of Carvacrol (CAS {cas_number}): {density} g/cm³")
            else:
                print("Density not available via PubChem for this compound.")
        else:
            print("Failed to fetch density data.")
    else:
        print(f"No CID found for CAS {cas_number}")
else:
    print("Failed to resolve CAS number.")
```
